```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import os
_project = os.environ.get("CHROMHMM_PROJECT_DIR", "chromHMM_project")
_model_dir = os.path.join(_project, "chromHMM_out", "model")
emission_file = os.path.join(_model_dir, "emissions_10.txt")
htmap_file = os.path.join(_project, "..", "output_figs", "emission10_heatmap.pdf")
os.makedirs(os.path.dirname(htmap_file), exist_ok=True)

df = pd.read_csv(emission_file, sep="\t", index_col=0)

fig, ax = plt.subplots(figsize=(3,4))
sns.heatmap(df, cmap="Purples", linewidths=0.7, linecolor="black", square=True, ax=ax)

# Ensure that all plot spines are visible
for spine in ax.spines.values():
    spine.set_visible(True)
    spine.set_linewidth(0.7)
    spine.set_color('black')

ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right')
ax.set_yticklabels(ax.get_yticklabels(), rotation=0)

plt.tight_layout()
# plt.show()

fig.savefig(htmap_file)
```
